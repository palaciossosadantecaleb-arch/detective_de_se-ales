```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv1D, MaxPooling1D, Flatten, Dense, Dropout
from tensorflow.keras.utils import to_categorical

# Parámetros de la señal
fs = 1000          # Frecuencia de muestreo (Hz)
T = 0.1            # Duración de la muestra de la señal (segundos)
t = np.linspace(0, T, int(fs * T), endpoint=False)
fc = 50            # Frecuencia portadora (Hz)

print("¡Entorno configurado y parámetros definidos exitosamente!")
```

```python
def generar_ask(t, fc):
    bits = np.random.randint(0, 2, len(t))
    return bits * np.sin(2 * np.pi * fc * t)

def generar_fsk(t, fc):
    bits = np.random.randint(0, 2, len(t))
    f1, f2 = fc - 15, fc + 15
    freqs = np.where(bits == 1, f1, f2)
    return np.sin(2 * np.pi * freqs * t)

def generar_psk(t, fc):
    bits = np.random.randint(0, 2, len(t))
    phase = np.where(bits == 1, 0, np.pi)
    return np.sin(2 * np.pi * fc * t + phase)

def generar_qam(t, fc):
    # Simulación de 16-QAM combinando amplitud y fase discretas
    amplitud = np.random.choice([0.5, 1.0, 1.5, 2.0], size=len(t))
    fase = np.random.choice([0, np.pi/2, np.pi, 3*np.pi/2], size=len(t))
    return amplitud * np.sin(2 * np.pi * fc * t + fase)
```

```python
def agregar_awgn(signal, snr_db):
    sig_power = np.mean(signal ** 2)
    snr_linear = 10 ** (snr_db / 10)
    noise_power = sig_power / snr_linear
    noise = np.random.normal(0, np.sqrt(noise_power), signal.shape)
    return signal + noise

muestras_por_clase = 1000
clases = ['ASK', 'FSK', 'PSK', 'QAM']
X = []
y = []

np.random.seed(42)

for etiqueta, clase in enumerate(clases):
    for _ in range(muestras_por_clase):
        if clase == 'ASK':
            sig = generar_ask(t, fc)
        elif clase == 'FSK':
            sig = generar_fsk(t, fc)
        elif clase == 'PSK':
            sig = generar_psk(t, fc)
        elif clase == 'QAM':
            sig = generar_qam(t, fc)

        # Añadir ruido aleatorio (SNR entre 0 y 25 dB)
        snr_aleatorio = np.random.uniform(0, 25)
        sig_ruidosa = agregar_awgn(sig, snr_aleatorio)

        X.append(sig_ruidosa)
        y.append(etiqueta)

X = np.array(X)
y = np.array(y)

# Expandir dimensiones para la red convolucional 1D: (muestras, puntos, canales)
X = np.expand_dims(X, axis=-1)
y_categorical = to_categorical(y, num_classes=4)

print(f"Dataset generado con éxito. Forma de X: {X.shape}, Forma de y: {y_categorical.shape}")
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y_categorical, test_size=0.2, random_state=42)

print(f"Muestras para entrenamiento: {X_train.shape[0]}")
print(f"Muestras para prueba (test): {X_test.shape[0]}")
```

```python
# Gráficas de las modulaciones
plt.figure(figsize=(12, 8))

ejemplos = [
    ("ASK", generar_ask(t, fc)),
    ("FSK", generar_fsk(t, fc)),
    ("PSK", generar_psk(t, fc)),
    ("16-QAM", generar_qam(t, fc))
]

for i, (nombre, senal_limpia) in enumerate(ejemplos):
    senal_ruido = agregar_awgn(senal_limpia, snr_db=10)

    plt.subplot(4, 2, 2*i + 1)
    plt.plot(t, senal_limpia, color='teal')
    plt.title(f"{nombre} - Señal Limpia")
    plt.grid(True, alpha=0.3)

    plt.subplot(4, 2, 2*i + 2)
    plt.plot(t, senal_ruido, color='crimson')
    plt.title(f"{nombre} - Con Ruido AWGN (10 dB)")
    plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv1D, MaxPooling1D, Flatten, Dense, Dropout

# 2.2 Diseñar el modelo con una arquitectura CNN 1D
model = Sequential([
    # Capa de entrada y primera convolución con activación ReLU
    Conv1D(filters=32, kernel_size=5, activation='relu', input_shape=(X.shape[1], 1)),
    MaxPooling1D(pool_size=2),

    # Segunda capa convolucional para extraer características más complejas
    Conv1D(filters=64, kernel_size=5, activation='relu'),
    MaxPooling1D(pool_size=2),

    # Aplanar los datos para conectarlos con la capa densa
    Flatten(),

    # Capa oculta densa
    Dense(64, activation='relu'),
    Dropout(0.3), # Previene el sobreajuste (overfitting)

    # Capa de salida con 4 neuronas (Softmax) para las 4 modulaciones (ASK, FSK, PSK, QAM)
    Dense(4, activation='softmax')
])

print("¡Modelo diseñado con éxito!")
model.summary()
```

```python
# 2.3 Compilar el modelo
model.compile(optimizer='adam',
              loss='categorical_crossentropy',
              metrics=['accuracy'])

print("¡Modelo compilado correctamente!")
```

```python
import matplotlib.pyplot as plt

# 2.4 Entrenar el modelo (.fit)
history = model.fit(X_train, y_train,
                    epochs=10,
                    batch_size=32,
                    validation_data=(X_test, y_test))

# Gráficas de monitoreo (Precisión y Pérdida)
plt.figure(figsize=(12, 4))

plt.subplot(1, 2, 1)
plt.plot(history.history['accuracy'], label='Entrenamiento', color='blue')
plt.plot(history.history['val_accuracy'], label='Validación', color='orange')
plt.title('Precisión (Accuracy) del Modelo')
plt.xlabel('Épocas')
plt.ylabel('Accuracy')
plt.legend()
plt.grid(True, alpha=0.3)

plt.subplot(1, 2, 2)
plt.plot(history.history['loss'], label='Entrenamiento', color='blue')
plt.plot(history.history['val_loss'], label='Validación', color='orange')
plt.title('Pérdida (Loss) del Modelo')
plt.xlabel('Épocas')
plt.ylabel('Loss')
plt.legend()
plt.grid(True, alpha=0.3)

plt.show()
```

```python
# 2.5 Guardar el modelo entrenado
model.save("detector_senales_cnn.h5")

print("¡Modelo exportado y guardado exitosamente como 'detector_senales_cnn.h5'!")
```

```python
from sklearn.metrics import confusion_matrix, classification_report, ConfusionMatrixDisplay
import numpy as np
import matplotlib.pyplot as plt

# 3.1 Predicciones sobre el conjunto de prueba
y_pred_probs = model.predict(X_test)
y_pred = np.argmax(y_pred_probs, axis=1)
y_true = np.argmax(y_test, axis=1)

clases = ['ASK', 'FSK', 'PSK', 'QAM']

# Generar y mostrar Matriz de Confusión
cm = confusion_matrix(y_true, y_pred)
disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=clases)

plt.figure(figsize=(7, 6))
disp.plot(cmap='Blues', values_format='d')
plt.title('Matriz de Confusión - Conjunto de Prueba')
plt.show()

print("\nReporte de clasificación:")
print(classification_report(y_true, y_pred, target_names=clases))
```

```python
# 3.2 y 3.3 Prueba de estrés con distintos niveles de SNR y gráfica de rendimiento
niveles_snr = [-5, 0, 5, 10, 15, 20, 25]
precisiones_snr = []

# Regeneramos señales limpias base para controlar exactamente el SNR de prueba
np.random.seed(123)
muestras_por_snr = 200  # 50 por cada clase por cada nivel de SNR

for snr in niveles_snr:
    X_snr = []
    y_snr = []
    for _ in range(muestras_por_snr // 4):
        for idx, clase in enumerate(clases):
            if clase == 'ASK':
                s = generar_ask(t, fc)
            elif clase == 'FSK':
                s = generar_fsk(t, fc)
            elif clase == 'PSK':
                s = generar_psk(t, fc)
            elif clase == 'QAM':
                s = generar_qam(t, fc)
            s_ruidosa = agregar_awgn(s, snr)
            X_snr.append(s_ruidosa)
            y_snr.append(idx)

    X_snr = np.expand_dims(np.array(X_snr), axis=-1)
    y_snr_cat = to_categorical(np.array(y_snr), num_classes=4)

    _, acc = model.evaluate(X_snr, y_snr_cat, verbose=0)
    precisiones_snr.append(acc * 100)
    print(f"SNR: {snr:>3} dB  -->  Precisión: {acc*100:.2f}%")

# Gráfica de rendimiento (Eje X: SNR, Eje Y: Precisión)
plt.figure(figsize=(8, 5))
plt.plot(niveles_snr, precisiones_snr, marker='o', linewidth=2, color='darkgreen')
plt.title('Rendimiento del Modelo vs. Nivel de Ruido (SNR)')
plt.xlabel('SNR (dB) [Menor = Más ruido, Mayor = Menos ruido]')
plt.ylabel('Precisión (%)')
plt.grid(True, alpha=0.3)
plt.ylim(0, 105)
plt.show()
```

```python
# 3.4 Precisión por modulación para identificar la más difícil
print("Precisión individual por modulación en el conjunto de prueba:")
for idx, nombre in enumerate(clases):
    mask = (y_true == idx)
    acc_clase = np.mean(y_pred[mask] == y_true[mask]) * 100
    errores_clase = np.sum(y_pred[mask] != y_true[mask])
    print(f"- {nombre:<4}: {acc_clase:.2f}% de acierto ({errores_clase} errores de {np.sum(mask)} muestras)")
```

```python
import numpy as np
import matplotlib.pyplot as plt

# 4.2 Demostración en vivo: clasificar una señal nueva al azar
clases = ['ASK', 'FSK', 'PSK', 'QAM']
indice_real = np.random.randint(0, len(clases))
mod_real = clases[indice_real]
snr_demo = np.random.uniform(2, 20)  # Ruido aleatorio para la prueba en vivo

# Generar señal nueva limpia
if mod_real == 'ASK':
    senal_limpia = generar_ask(t, fc)
elif mod_real == 'FSK':
    senal_limpia = generar_fsk(t, fc)
elif mod_real == 'PSK':
    senal_limpia = generar_psk(t, fc)
elif mod_real == 'QAM':
    senal_limpia = generar_qam(t, fc)

# Aplicar ruido AWGN
senal_nueva = agregar_awgn(senal_limpia, snr_demo)

# Preparar entrada con forma (1, longitud_senal, 1)
entrada_modelo = np.expand_dims(senal_nueva, axis=(0, -1))

# Predicción del modelo
probabilidades = model.predict(entrada_modelo, verbose=0)[0]
indice_pred = np.argmax(probabilidades)
mod_pred = clases[indice_pred]

print(f"Modulación REAL:        {mod_real} (SNR de prueba: {snr_demo:.1f} dB)")
print(f"Predicción de la IA:    {mod_pred} (Confianza: {probabilidades[indice_pred]*100:.2f}%)\n")
for nombre_c, prob in zip(clases, probabilidades):
    print(f"  - Probabilidad {nombre_c:<4}: {prob*100:.2f}%")

# Graficar la señal nueva clasificada
plt.figure(figsize=(10, 4))
plt.plot(t, senal_nueva, color='purple')
plt.title(f"Señal Nueva (SNR ~ {snr_demo:.1f} dB) | Real: {mod_real} | Predicción IA: {mod_pred}")
plt.xlabel("Tiempo (s)")
plt.ylabel("Amplitud")
plt.grid(True, alpha=0.3)
plt.show()
```
